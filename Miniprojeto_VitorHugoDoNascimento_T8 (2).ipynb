2. Limpeza justificada

As decisões da limpeza será remover as linhas com #N/D e colunas que representam NaN.

```python
# Corrigindo espaços e padronizando


def tratar_dados(tabela):
  tratado = tabela.copy()

  for coluna in ["PR_ID", "PR_NOME", "CL_GENERO", "CL_SEG"]:
    tratado[coluna] = (
        tratado[coluna]
        .astype("string")
        .str.strip()
        .replace("", pd.NA)
        .str.title()
    )

  return tratado

df = tratar_dados(df)
display(df)
```

```python
# Retirando colunas unnamed
df = df.drop(columns=['Unnamed: 10', 'Unnamed: 11', 'Unnamed: 12', 'Unnamed: 13'])

display(df)
```

```python
df = df[df["PR_CAT"] != "#N/D"]
display(df)
```

## **3. Funções e Organização**

```python
# Comparação entre tabelas

def gerar_resumo(tabela):
    return {
        "linhas totais": tabela.shape[0],
        "colunas totais": tabela.shape[1]
    }
resumo_original = gerar_resumo(df_original)
resumo_tratado = gerar_resumo(df)
print("Linhas antes: ", resumo_original["linhas totais"])
print("Linhas depois: ", resumo_tratado["linhas totais"])
print("Colunas antes: ", resumo_original["colunas totais"])
print("Colunas depois: ", resumo_tratado["colunas totais"])

print(
    "Duplicatas restantes: ",
    df.duplicated().sum()
)

print("\nValores nulos por coluna após o tratamento:")
display(df.isna().sum())
```

Notamos que após o tratamento foram 3650 linhas cortadas da tabela original, sendo valores não definidos, também foram identificadas a diminuição de 14 para 10 colunas, sendo as 4 colunas sem nome e sem dados cortadas.

## **4. Análise**

Iremos realizar as seguintes análises:
Quantidade de clientes na base, de itens, lojas, Média de filhos, minimo, máximo e quantidade de clientes com pelo menos um filho

```python
import pandas as pd

media_filhos = df.drop_duplicates(subset=["CL_ID"])["CL_FHL"].mean()
qtd_clientes_com_filhos = (df.drop_duplicates(subset=["CL_ID"])["CL_FHL"] >= 1).sum()
print("Quantidade de clientes na base: ", df["CL_ID"].nunique())
print("Quantidade de itens na base: ", df["PR_ID"].nunique())
print("Quantidade de lojas na base: ", df["CO_ID"].nunique())
print("Média de filhos por cliente: ", media_filhos)
print("Menor número de filhos: ", df["CL_FHL"].min())
print("Maior número de filhos: ", df["CL_FHL"].max())
print("Quantidade de clientes com pelo menos um filho que realizaram a compra: ", qtd_clientes_com_filhos)
```

É notado que a média de filhos não é muito alto, sendo 1,136, tendo mais da metade dos clientes com nenhum filho. O cliente que tem mais filhos tem 4, com menos é 0. Tem 1000 clientes diferentes na base, 228 itens e 18469 lojas.

Farei a análise entre os principais produtos mais e menos vendidos por categoria.

```python
top_5_por_categoria = (
    df.groupby(["PR_CAT", "PR_NOME"])["CO_ID"]
    .nunique()
    .reset_index(name="total")
    .sort_values(by=["PR_CAT", "total"], ascending=[True, False])
    .groupby("PR_CAT")
    .head(5) )

display(top_5_por_categoria)
```

```python
last_5_por_categoria = (
    df.groupby(["PR_CAT", "PR_NOME"])["CO_ID"]
    .nunique()
    .reset_index(name="total")
    .sort_values(by=["PR_CAT", "total"], ascending=[True, False])
    .groupby("PR_CAT")
    .tail(5) )

display(last_5_por_categoria)
```

Com essa análise, podemos observar que a categoria de acessórios, tem apenas dois produtos. A categoria de bebidas tem apenas 6 produtos, sendo 5 refrigerantes. Categoria Pet tem apenas 5 produtos, onde é notado uma diferença muito grande entre as vendas de rações para cachorros ou gatos em relação aos de passáros. A diferença de vendas entre a Cera, produto mais vendido na categoria de limpeza, para sabão em pó, terceiro menos vendido, é de menos de 200 vendas. Sendo uma categoria bem equilibrada. A categoria de higiene também é outra equilibrada, apenas absorvente com uma menor frequência. O presunto é o produto mais vendido, com sobras, no setor de alimentos.
